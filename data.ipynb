```python
import pandas as pd

df = pd.read_csv('data/raw_data.csv')
```

```python
# 학과명 채우기
df.loc[0:3, '학과명'] = '컴퓨터공학전공'
df.loc[4:15, '학과명'] = '소프트웨어전공'
df.loc[16:21, '학과명'] = '정보통신공학전공'
df.loc[23:31, '학과명'] = '인공지능전공'
df.loc[32:48, '학과명'] = '소프트웨어융합전공'
df.loc[49:51, '학과명'] = '빅데이터응용전공'
df.loc[51:, '학과명'] = '인공지능전공'
```

```python
#학부명 채우기
df.loc[0:31, '학부명'] = 'IT융합자율학부'
df.loc[32:48, '학부명'] = '소프트웨어융합학부'
df.loc[49:, '학부명'] = '미래융합학부'
```

```python
#이수구분 none값 지우기
df = df[df['이수구분'].notna()]
```

```python
#강의평가 이후 컬럼 삭제
df = df.loc[:, :'강의평가']
```

```python
#인덱스 초기화
df = df.reset_index(drop=True)
```

```python
df
```

```python
#선수과목과수강요건 none값 '없음'으로 채우기
df['선수과목과수강요건'] = df['선수과목과수강요건'].fillna('없음')
```

```python
#별점 none값 '없음'으로 채우기
df['별점'] = df['별점'].fillna('없음')
```

```python
df['강의평가'] = df['강의평가'].fillna('없음')
```

```python
df['학습내용'] = df['학습내용'].fillna('없음')
```

```python
df
```

```python
#강의평가 데이터를\\n기준으로 나누고 (1)(2)(3)이 포함되면 나눠진 문장 단위로강의평가_낮음 컬럼에 추가
df['강의평가_낮음'] = df['강의평가'].apply(lambda x: '\n'.join([line for line in x.split('\n') if '(1)' in line or '(2)' in line or '(3)' in line]))
df['강의평가_높음'] = df['강의평가'].apply(lambda x: '\n'.join([line for line in x.split('\n') if '(4)' in line or '(5)' in line]))

print("강의평가: \n", df['강의평가'][4])
print("---------------------------------------------------------------------")
print("강의평가 낮음: \n", df['강의평가_낮음'][4])
print("---------------------------------------------------------------------")
print("강의평가 높음: \n", df['강의평가_높음'][4])
```

```python
df['강의평가'].value_counts()
```

```python
#강의평가 x 컬럼 없음으로 교체
df['강의평가'] = df['강의평가'].replace('x', '없음')
df['강의평가'].value_counts()
```

```python
#df 저장
df.to_csv('data/cleaned_data.csv', index=False)
```
